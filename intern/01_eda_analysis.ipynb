```python
"""
Titanic EDA - Complete Analysis
Run this as a Jupyter notebook or Python script
"""

import sys
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Add src to path
sys.path.append(os.path.dirname(os.path.abspath(__file__)))

from src.data_loader import load_titanic_data, get_data_info, clean_data
from src.visualization import *

# Set up visualization style
setup_style()

print("="*80)
print("TITANIC EXPLORATORY DATA ANALYSIS (EDA)")
print("="*80)

# 1. Load and Prepare Data
print("\n📊 STEP 1: DATA LOADING AND INITIAL INSPECTION")
print("-"*50)

df = load_titanic_data()
info = get_data_info(df)

print(f"\nDataset Shape: {info['shape'][0]} rows, {info['shape'][1]} columns")
print(f"\nFirst 5 rows:")
print(df.head())
print(f"\nDataset Info:")
print(df.info())

# 2. Data Cleaning
print("\n🧹 STEP 2: DATA CLEANING")
print("-"*50)

df_clean = clean_data(df)
print(f"Missing values after cleaning: {df_clean.isnull().sum().sum()}")

# 3. Descriptive Statistics
print("\n📈 STEP 3: DESCRIPTIVE STATISTICS")
print("-"*50)

print("\nNumerical Features Statistics:")
print(df_clean.describe())
print("\nCategorical Features:")
print(df_clean.describe(include=['object']))

# 4. Univariate Analysis
print("\n📊 STEP 4: UNIVARIATE ANALYSIS")
print("-"*50)

# Survival rate
survival_rate = df_clean['Survived'].mean() * 100
print(f"\nOverall Survival Rate: {survival_rate:.1f}%")

# Gender distribution
gender_counts = df_clean['Sex'].value_counts()
print(f"\nGender Distribution:\n{gender_counts}")

# Class distribution
class_counts = df_clean['Pclass'].value_counts()
print(f"\nPassenger Class Distribution:\n{class_counts}")

# Age statistics
print(f"\nAge Statistics:")
print(f"  Mean: {df_clean['Age'].mean():.1f} years")
print(f"  Median: {df_clean['Age'].median():.1f} years")
print(f"  Range: {df_clean['Age'].min()} - {df_clean['Age'].max()} years")

# Fare statistics
print(f"\nFare Statistics:")
print(f"  Mean: ${df_clean['Fare'].mean():.2f}")
print(f"  Median: ${df_clean['Fare'].median():.2f}")
print(f"  Range: ${df_clean['Fare'].min():.2f} - ${df_clean['Fare'].max():.2f}")

# 5. Bivariate Analysis
print("\n🔍 STEP 5: BIVARIATE ANALYSIS")
print("-"*50)

# Survival by gender
survival_by_gender = df_clean.groupby('Sex')['Survived'].mean() * 100
print(f"\nSurvival Rate by Gender:")
print(f"  Female: {survival_by_gender['female']:.1f}%")
print(f"  Male: {survival_by_gender['male']:.1f}%")
print(f"  Difference: {survival_by_gender['female'] - survival_by_gender['male']:.1f}%")

# Survival by class
survival_by_class = df_clean.groupby('Pclass')['Survived'].mean() * 100
print(f"\nSurvival Rate by Passenger Class:")
for i in [1, 2, 3]:
    print(f"  Class {i}: {survival_by_class[i]:.1f}%")

# Survival by age group
if 'AgeGroup' in df_clean.columns:
    survival_by_age = df_clean.groupby('AgeGroup')['Survived'].mean() * 100
    print(f"\nSurvival Rate by Age Group:")
    for age_group, rate in survival_by_age.items():
        print(f"  {age_group}: {rate:.1f}%")

# Survival by family size
survival_by_family = df_clean.groupby('FamilySize')['Survived'].mean() * 100
print(f"\nSurvival Rate by Family Size:")
for size, rate in survival_by_family.items():
    print(f"  Size {size}: {rate:.1f}%")

# 6. Correlation Analysis
print("\n📉 STEP 6: CORRELATION ANALYSIS")
print("-"*50)

numeric_cols = ['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare', 'FamilySize']
correlation_matrix = df_clean[numeric_cols].corr()

print("\nCorrelation with Survival:")
survival_corr = correlation_matrix['Survived'].sort_values(ascending=False)
for feature, corr in survival_corr.items():
    if feature != 'Survived':
        print(f"  {feature}: {corr:.3f}")

# 7. Generate Visualizations
print("\n🎨 STEP 7: GENERATING VISUALIZATIONS")
print("-"*50)

# Create images directory
if not os.path.exists('images'):
    os.makedirs('images')

# Generate and save plots
print("Generating survival rate plot...")
fig1 = plot_survival_rate(df_clean)
fig1.savefig('images/survival_rate.png', dpi=100, bbox_inches='tight')
plt.close(fig1)

print("Generating gender analysis plot...")
fig2 = plot_survival_by_gender(df_clean)
fig2.savefig('images/survival_by_gender.png', dpi=100, bbox_inches='tight')
plt.close(fig2)

print("Generating class analysis plot...")
fig3 = plot_survival_by_class(df_clean)
fig3.savefig('images/survival_by_class.png', dpi=100, bbox_inches='tight')
plt.close(fig3)

print("Generating age distribution plot...")
fig4 = plot_age_distribution(df_clean)
fig4.savefig('images/age_distribution.png', dpi=100, bbox_inches='tight')
plt.close(fig4)

print("Generating fare analysis plot...")
fig5 = plot_fare_analysis(df_clean)
fig5.savefig('images/fare_analysis.png', dpi=100, bbox_inches='tight')
plt.close(fig5)

print("Generating correlation heatmap...")
fig6 = plot_correlation_heatmap(df_clean)
fig6.savefig('images/correlation_heatmap.png', dpi=100, bbox_inches='tight')
plt.close(fig6)

print("Generating family impact plot...")
fig7 = plot_family_impact(df_clean)
fig7.savefig('images/family_impact.png', dpi=100, bbox_inches='tight')
plt.close(fig7)

print("Generating embarkation port plot...")
fig8 = plot_embarkation_port(df_clean)
fig8.savefig('images/embarkation_port.png', dpi=100, bbox_inches='tight')
plt.close(fig8)

print("Generating survival heatmap...")
fig9 = plot_survival_heatmap(df_clean)
fig9.savefig('images/survival_heatmap.png', dpi=100, bbox_inches='tight')
plt.close(fig9)

# 8. Key Insights
print("\n💡 STEP 8: KEY INSIGHTS AND FINDINGS")
print("-"*50)

insights = {
    "Gender Impact": "Women had significantly higher survival rates (74%) compared to men (19%)",
    "Class Privilege": "First-class passengers (63% survival) had much better chances than third-class (24%)",
    "Age Factor": "Children under 12 had higher survival rates than adults",
    "Economic Status": "Higher ticket fares correlated strongly with survival (correlation: 0.26)",
    "Family Size": "Small families (2-4 members) had optimal survival rates",
    "Embarkation": "Passengers from Cherbourg had higher survival rates"
}

for key, value in insights.items():
    print(f"\n• {key}:")
    print(f"  → {value}")

# 9. Recommendations
print("\n📋 STEP 9: RECOMMENDATIONS")
print("-"*50)

recommendations = [
    "Prioritize vulnerable groups (women, children, elderly) in evacuation procedures",
    "First-class passengers should assist lower-class passengers in emergencies",
    "Family reunification protocols should be established before boarding",
    "Training should emphasize rapid, orderly evacuation regardless of ticket class"
]

for i, rec in enumerate(recommendations, 1):
    print(f"{i}. {rec}")

print("\n" + "="*80)
print("✅ EDA COMPLETED SUCCESSFULLY!")
print(f"📁 Visualizations saved in 'images' directory")
print("="*80)

# Optional: Interactive summary
print("\n📊 SUMMARY STATISTICS TABLE")
print("-"*50)

summary_table = pd.DataFrame({
    'Metric': ['Total Passengers', 'Survival Rate', 'Female Survival', 'Male Survival',
               'First Class Survival', 'Second Class Survival', 'Third Class Survival',
               'Average Age', 'Average Fare', 'Children Survival Rate'],
    'Value': [len(df_clean), f"{survival_rate:.1f}%", f"{survival_by_gender['female']:.1f}%",
              f"{survival_by_gender['male']:.1f}%", f"{survival_by_class[1]:.1f}%",
              f"{survival_by_class[2]:.1f}%", f"{survival_by_class[3]:.1f}%",
              f"{df_clean['Age'].mean():.1f} years", f"${df_clean['Fare'].mean():.2f}",
              f"{survival_by_age['Child']:.1f}%" if 'AgeGroup' in df_clean.columns else "N/A"]
})

print(summary_table.to_string(index=False))
```
